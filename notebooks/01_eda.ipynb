```python
# Calculate implied probabilities from Bet365 odds
df_combined["B365_implied_H"] = 1 / df_combined["B365H"]
df_combined["B365_implied_D"] = 1 / df_combined["B365D"]
df_combined["B365_implied_A"] = 1 / df_combined["B365A"]

# Sum of implied probabilities (should be > 1 due to bookmaker margin)
df_combined["B365_overround"] = (
    df_combined["B365_implied_H"] + 
    df_combined["B365_implied_D"] + 
    df_combined["B365_implied_A"]
)

print(f"Mean overround: {df_combined['B365_overround'].mean():.4f}")
print(f"Min overround:  {df_combined['B365_overround'].min():.4f}")
print(f"Max overround:  {df_combined['B365_overround'].max():.4f}")
```

```python
bookmakers = {
    "B365": ("B365H", "B365D", "B365A"),
    "BW": ("BWH", "BWD", "BWA"),
    "IW": ("IWH", "IWD", "IWA"),
    "PS": ("PSH", "PSD", "PSA"),
    "WH": ("WHH", "WHD", "WHA"),
}

for bk, (h_col, d_col, a_col) in bookmakers.items():
    df_combined[f"{bk}_implied_H"] = 1 / df_combined[h_col]
    df_combined[f"{bk}_implied_D"] = 1 / df_combined[d_col]
    df_combined[f"{bk}_implied_A"] = 1 / df_combined[a_col]
    df_combined[f"{bk}_overround"] = (
        df_combined[f"{bk}_implied_H"]
        + df_combined[f"{bk}_implied_D"]
        + df_combined[f"{bk}_implied_A"]
    )

for bk in bookmakers:
    col = f"{bk}_overround"
    print(
        f"{bk} mean: {df_combined[col].mean():.4f}, "
        f"min: {df_combined[col].min():.4f}, "
        f"max: {df_combined[col].max():.4f}"
    )
```

```python
# Defragment first
df_combined = df_combined.copy()

# Build all new columns at once
new_cols = {}
for bk in bookmakers:
    new_cols[f"{bk}_norm_H"] = df_combined[f"{bk}_implied_H"] / df_combined[f"{bk}_overround"]
    new_cols[f"{bk}_norm_D"] = df_combined[f"{bk}_implied_D"] / df_combined[f"{bk}_overround"]
    new_cols[f"{bk}_norm_A"] = df_combined[f"{bk}_implied_A"] / df_combined[f"{bk}_overround"]

# Add all columns at once
df_combined = pd.concat([df_combined, pd.DataFrame(new_cols, index=df_combined.index)], axis=1)
```

```python
for bk in bookmakers:
    mean_sum = float((df_combined[f"{bk}_norm_H"] + 
                df_combined[f"{bk}_norm_D"] + 
                df_combined[f"{bk}_norm_A"]).mean(skipna=True))
    print(f"{bk}: sum of normalized probs = {mean_sum:.6f}")
```

```python
for bk, (h_col, d_col, a_col) in bookmakers.items():
    nulls = df_combined[[h_col, d_col, a_col]].isnull().any(axis=1).sum()
    print(f"{bk}: {nulls} partidos con cuotas missing ({nulls/len(df_combined):.1%})")
    
```

```python
# Find all bookmaker columns that have H, D, A variants
import re

all_cols = df_combined.columns.tolist()

# Find all columns ending in H, D, A that look like odds
bookmaker_h_cols = [c for c in all_cols if c.endswith('H') and not c.startswith('Bb') 
                    and c not in ['FTHG', 'HTAG', 'HST', 'HC', 'HF', 'HY', 'HR', 'HS', 'HTHG']]

# For each H column, check if D and A variants exist
all_bookmakers = {}
for h_col in bookmaker_h_cols:
    prefix = h_col[:-1]
    d_col = f"{prefix}D"
    a_col = f"{prefix}A"
    if d_col in all_cols and a_col in all_cols:
        total = len(df_combined)
        missing = df_combined[[h_col, d_col, a_col]].isnull().any(axis=1).sum()
        coverage = (total - missing) / total
        all_bookmakers[prefix] = coverage

# Sort by coverage
sorted_bks = sorted(all_bookmakers.items(), key=lambda x: x[1], reverse=True)
for bk, coverage in sorted_bks:
    print(f"{bk:15} {coverage:.1%} coverage ({int(coverage*4940)} partidos)")
    
```

```python
# Convert FTR to one-hot encoding
df_combined["real_H"] = (df_combined["FTR"] == "H").astype(int)
df_combined["real_D"] = (df_combined["FTR"] == "D").astype(int)
df_combined["real_A"] = (df_combined["FTR"] == "A").astype(int)

# Verify
print(df_combined[["FTR", "real_H", "real_D", "real_A"]].head(10))
```

```python
from sklearn.metrics import log_loss, brier_score_loss
import numpy as np

y_true = df_combined[["real_H", "real_D", "real_A"]].values
bookmakers_final = ["B365", "BW", "PS", "WH"]

print(f"{'Casa':10} {'Log Loss':>10} {'Brier Score':>12} {'Accuracy':>10}")
print("-" * 45)

for bk in bookmakers_final:
    mask = df_combined[[f"{bk}_norm_H", f"{bk}_norm_D", f"{bk}_norm_A"]].notna().all(axis=1)
    y_true_bk = y_true[mask]
    y_pred = df_combined.loc[mask, [f"{bk}_norm_H", f"{bk}_norm_D", f"{bk}_norm_A"]].values
    
    ll = log_loss(y_true_bk, y_pred)
    bs = np.mean([brier_score_loss(y_true_bk[:, i], y_pred[:, i]) for i in range(3)])
    
    predicted_result = np.argmax(y_pred, axis=1)
    actual_result = np.argmax(y_true_bk, axis=1)
    acc = (predicted_result == actual_result).mean()
    
    print(f"{bk:10} {ll:>10.4f} {bs:>12.4f} {acc:>10.1%}")
```

```python
# Analyze odds movement for Pinnacle
ps_both = df_combined[df_combined[["PSH", "PSD", "PSA", "PSCH", "PSCD", "PSCA"]].notna().all(axis=1)].copy()

# Calculate implied probs for opening and closing
for result, open_col, close_col in [("H", "PSH", "PSCH"), ("D", "PSD", "PSCD"), ("A", "PSA", "PSCA")]:
    ps_both[f"open_prob_{result}"] = 1 / ps_both[open_col]
    ps_both[f"close_prob_{result}"] = 1 / ps_both[close_col]
    ps_both[f"movement_{result}"] = ps_both[f"close_prob_{result}"] - ps_both[f"open_prob_{result}"]

# Average movement per result
print("Media de movimiento de probabilidad (cierre - apertura):")
print(f"  Local:    {ps_both['movement_H'].mean():+.4f}")
print(f"  Empate:   {ps_both['movement_D'].mean():+.4f}")
print(f"  Visitante:{ps_both['movement_A'].mean():+.4f}")

# How often does the market move significantly (>2%)
threshold = 0.02
for result in ["H", "D", "A"]:
    big_move = (ps_both[f"movement_{result}"].abs() > threshold).mean()
    print(f"\nMovimiento >2% en {result}: {big_move:.1%} de partidos")
```

```python
from sklearn.metrics import log_loss
import numpy as np

# Normalize opening odds
ps_both["open_sum"] = ps_both["open_prob_H"] + ps_both["open_prob_D"] + ps_both["open_prob_A"]
ps_both["open_norm_H"] = ps_both["open_prob_H"] / ps_both["open_sum"]
ps_both["open_norm_D"] = ps_both["open_prob_D"] / ps_both["open_sum"]
ps_both["open_norm_A"] = ps_both["open_prob_A"] / ps_both["open_sum"]

# Normalize closing odds
ps_both["close_sum"] = ps_both["close_prob_H"] + ps_both["close_prob_D"] + ps_both["close_prob_A"]
ps_both["close_norm_H"] = ps_both["close_prob_H"] / ps_both["close_sum"]
ps_both["close_norm_D"] = ps_both["close_prob_D"] / ps_both["close_sum"]
ps_both["close_norm_A"] = ps_both["close_prob_A"] / ps_both["close_sum"]

# Results
y_true = ps_both[["real_H", "real_D", "real_A"]].values
y_open = ps_both[["open_norm_H", "open_norm_D", "open_norm_A"]].values
y_close = ps_both[["close_norm_H", "close_norm_D", "close_norm_A"]].values

ll_open = log_loss(y_true, y_open)
ll_close = log_loss(y_true, y_close)

acc_open = (np.argmax(y_open, axis=1) == np.argmax(y_true, axis=1)).mean()
acc_close = (np.argmax(y_close, axis=1) == np.argmax(y_true, axis=1)).mean()

print(f"{'':15} {'Log Loss':>10} {'Accuracy':>10}")
print(f"{'PS Apertura':15} {ll_open:>10.4f} {acc_open:>10.1%}")
print(f"{'PS Cierre':15} {ll_close:>10.4f} {acc_close:>10.1%}")
print(f"\nMejora Log Loss: {ll_open - ll_close:+.4f}")
```

```python
bookmakers_oc = {
    "B365": {"open": ("B365H", "B365D", "B365A"), "close": ("B365CH", "B365CD", "B365CA")},
    "BW":   {"open": ("BWH", "BWD", "BWA"),       "close": ("BWCH", "BWCD", "BWCA")},
    "PS":   {"open": ("PSH", "PSD", "PSA"),        "close": ("PSCH", "PSCD", "PSCA")},
    "WH":   {"open": ("WHH", "WHD", "WHA"),        "close": ("WHCH", "WHCD", "WHCA")},
}

for bk, cols in bookmakers_oc.items():
    open_cols = cols["open"]   
    close_cols = cols["close"]

    # Analyze odds movement for all bookmakers with both open and close bets
    all_cols = list(open_cols) + list(close_cols)
    bk_both = df_combined[df_combined[all_cols].notna().all(axis=1)].copy()

    # Calculate implied probs for opening and closing
    for result, open_col, close_col in zip(["H", "D", "A"], open_cols, close_cols): #Empareja los datos de varias listas en paralelo
        bk_both[f"open_prob_{result}"] = 1 / bk_both[open_col]
        bk_both[f"close_prob_{result}"] = 1 / bk_both[close_col]
        bk_both[f"movement_{result}"] = bk_both[f"close_prob_{result}"] - bk_both[f"open_prob_{result}"]

    # Average movement per result
    print("\nMedia de movimiento de probabilidad (cierre - apertura):")
    print(f"--- {bk} ({len(bk_both)} partidos) ---")
    print(f"  Local:    {bk_both['movement_H'].mean():+.4f}")
    print(f"  Empate:   {bk_both['movement_D'].mean():+.4f}")
    print(f"  Visitante:{bk_both['movement_A'].mean():+.4f}")

    # How often does the market move significantly (>2%)
    threshold = 0.02
    print(f"\nMovimiento >2% en {bk} ({len(bk_both)} partidos):")
    for result in ["H", "D", "A"]:
        big_move = (bk_both[f"movement_{result}"].abs() > threshold).mean()
        print(f"Movimiento >2% en {result}: {big_move:.1%} de partidos")

    # How often does the market move significantly (>2%)
    threshold = 0.10
    print(f"\nMovimiento >10% en {bk} ({len(bk_both)} partidos):")
    for result in ["H", "D", "A"]:
        big_move = (bk_both[f"movement_{result}"].abs() > threshold).mean()
        print(f"Movimiento >10% en {result}: {big_move:.1%} de partidos")
```

```python
from sklearn.metrics import log_loss, brier_score_loss
import numpy as np

for bk, cols in bookmakers_oc.items():
    # 1. Extraer columnas
    open_cols = cols["open"]   
    close_cols = cols["close"]

    # 2. Filtrar partidos con datos completos
    all_cols = list(open_cols) + list(close_cols)
    bk_both = df_combined[df_combined[all_cols].notna().all(axis=1)].copy()

    # 3. Calcular probabilidades implícitas
    for result, open_col, close_col in zip(["H", "D", "A"], open_cols, close_cols):
        bk_both[f"open_prob_{result}"] = 1 / bk_both[open_col]
        bk_both[f"close_prob_{result}"] = 1 / bk_both[close_col]

    # 4. Normalizar ← rellena tú
    bk_both["open_sum"] = bk_both["open_prob_H"] + bk_both["open_prob_D"] + bk_both["open_prob_A"]
    bk_both["open_norm_H"] = bk_both["open_prob_H"] / bk_both["open_sum"]
    bk_both["open_norm_D"] = bk_both["open_prob_D"] / bk_both["open_sum"]
    bk_both["open_norm_A"] = bk_both["open_prob_A"] / bk_both["open_sum"]
    bk_both["close_sum"] = bk_both["close_prob_H"] + bk_both["close_prob_D"] + bk_both["close_prob_A"]
    bk_both["close_norm_H"] = bk_both["close_prob_H"] / bk_both["close_sum"]
    bk_both["close_norm_D"] = bk_both["close_prob_D"] / bk_both["close_sum"]
    bk_both["close_norm_A"] = bk_both["close_prob_A"] / bk_both["close_sum"]

    # 5. Calcular métricas ← rellena tú
    y_true = bk_both[["real_H", "real_D", "real_A"]].values
    y_open = bk_both[["open_norm_H", "open_norm_D", "open_norm_A"]].values
    y_close = bk_both[["close_norm_H", "close_norm_D", "close_norm_A"]].values

    ll_open = log_loss(y_true, y_open)
    ll_close = log_loss(y_true, y_close)
    bs_open = np.mean([brier_score_loss(y_true[:, i], y_open[:, i]) for i in range(3)])
    bs_close = np.mean([brier_score_loss(y_true[:, i], y_close[:, i]) for i in range(3)])
    acc_open = (np.argmax(y_open, axis=1) == np.argmax(y_true, axis=1)).mean()
    acc_close = (np.argmax(y_close, axis=1) == np.argmax(y_true, axis=1)).mean()

    # 6. Imprimir
    print(f"\n--- {bk} ({len(bk_both)} partidos) ---")
    print(f"{'':15} {'Log Loss':>10} {'Brier':>8} {'Accuracy':>10}")
    print(f"{'Apertura':15} {ll_open:>10.4f} {bs_open:>8.4f} {acc_open:>10.1%}")
    print(f"{'Cierre':15} {ll_close:>10.4f} {bs_close:>8.4f} {acc_close:>10.1%}")
    print(f"Mejora LL: {ll_open - ll_close:+.4f}  Mejora BS: {bs_open - bs_close:+.4f}")
```
