```python
# Get all CSV files matching the pattern
csv_files = sorted(PROJECT_ROOT.glob("data/raw/laliga/SP1_*.csv"))

# Load all dataframes
dfs = [pd.read_csv(file) for file in csv_files]

# Check if all have the same columns
columns_match = all(df.columns.equals(dfs[0].columns) for df in dfs)
print(f"All dataframes have same columns: {columns_match}")

if not columns_match:
    print("\nColumns per file:")
for file, df in zip(csv_files, dfs):
    missing = set(dfs[0].columns) - set(df.columns)
    extra = set(df.columns) - set(dfs[0].columns)
    if missing or extra:
        print(f"{file.name}: missing={missing}, extra={extra}")

# Concatenate all dataframes
df_combined = pd.concat(dfs, ignore_index=True)
print(f"\nCombined shape: {df_combined.shape}")
print(f"Date range: {df_combined['Date'].min()} to {df_combined['Date'].max()}")
```

```python
df_combined["Date"] = pd.to_datetime(
    df_combined["Date"], 
    dayfirst=True
)

print(df_combined["Date"].dtype)
print(f"Date range: {df_combined['Date'].min()} to {df_combined['Date'].max()}")
```

```python
# Home advantage analysis
total = len(df_combined)
home_wins = (df_combined["FTR"] == "H").sum()
draws = (df_combined["FTR"] == "D").sum()
away_wins = (df_combined["FTR"] == "A").sum()

print(f"Home wins: {home_wins/total:.1%}")
print(f"Draws:     {draws/total:.1%}")
print(f"Away wins: {away_wins/total:.1%}")
```

```python
# Calculate implied probabilities from Bet365 odds
df_combined["B365_implied_H"] = 1 / df_combined["B365H"]
df_combined["B365_implied_D"] = 1 / df_combined["B365D"]
df_combined["B365_implied_A"] = 1 / df_combined["B365A"]

# Sum of implied probabilities (should be > 1 due to bookmaker margin)
df_combined["B365_overround"] = (
    df_combined["B365_implied_H"] + 
    df_combined["B365_implied_D"] + 
    df_combined["B365_implied_A"]
)

print(f"Mean overround: {df_combined['B365_overround'].mean():.4f}")
print(f"Min overround:  {df_combined['B365_overround'].min():.4f}")
print(f"Max overround:  {df_combined['B365_overround'].max():.4f}")
```

```python
bookmakers = {
    "B365": ("B365H", "B365D", "B365A"),
    "BW": ("BWH", "BWD", "BWA"),
    "IW": ("IWH", "IWD", "IWA"),
    "PS": ("PSH", "PSD", "PSA"),
    "WH": ("WHH", "WHD", "WHA"),
}

for bk, (h_col, d_col, a_col) in bookmakers.items():
    df_combined[f"{bk}_implied_H"] = 1 / df_combined[h_col]
    df_combined[f"{bk}_implied_D"] = 1 / df_combined[d_col]
    df_combined[f"{bk}_implied_A"] = 1 / df_combined[a_col]
    df_combined[f"{bk}_overround"] = (
        df_combined[f"{bk}_implied_H"]
        + df_combined[f"{bk}_implied_D"]
        + df_combined[f"{bk}_implied_A"]
    )

for bk in bookmakers:
    col = f"{bk}_overround"
    print(
        f"{bk} mean: {df_combined[col].mean():.4f}, "
        f"min: {df_combined[col].min():.4f}, "
        f"max: {df_combined[col].max():.4f}"
    )
```
